## Шаг 1. Загрузка данных

Загрузите файл и посмотрите на него.

```python
import pandas as pd

df = pd.read_csv("sales.csv")
df.head()
df.info()
```

## Шаг 2. Очистка

Задание: уберите дубликаты, заполните пропуски в сумме нулями, проверьте типы.

```python
# TODO: напишите (или попросите агента) очистку
# - df.drop_duplicates()
# - df["сумма"] = df["сумма"].fillna(0)
# - df.dtypes
df = df.drop_duplicates()
df["сумма"] = pd.to_numeric(df["сумма"], errors="coerce").fillna(0)
print("Дубликатов после очистки:", df.duplicated().sum())
print("Пропусков:", df.isna().sum().sum())
df.head()
```

## Шаг 3. Итоги по регионам

Посчитайте сумму и количество продаж по каждому региону.

```python
by_region = df.groupby("регион")["сумма"].agg(["sum", "count", "mean"])
by_region = by_region.sort_values("sum", ascending=False)
by_region
```

## Шаг 4. Контрольная сумма

Проверьте: сумма по регионам должна равняться общей сумме файла.

```python
total = df["сумма"].sum()
by_region_sum = by_region["sum"].sum()
print("Общая сумма:", total)
print("Сумма по регионам:", by_region_sum)
print("Сходятся?", abs(total - by_region_sum) < 1e-6)
```
