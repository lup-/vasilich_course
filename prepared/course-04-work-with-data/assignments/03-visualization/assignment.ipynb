## Шаг 1. Читаем данные

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales.csv")
df["дата"] = pd.to_datetime(df["дата"])
df["месяц"] = df["дата"].dt.to_period("M").astype(str)
df.head()
```

## Шаг 2. Линейный график продаж по месяцам

```python
monthly = df.groupby("месяц")["сумма"].sum()

plt.figure(figsize=(10, 5))
plt.plot(monthly.index, monthly.values, marker="o")
plt.title("Продажи по месяцам")
plt.xlabel("Месяц")
plt.ylabel("Сумма, руб.")
plt.grid(True)
plt.show()
```

## Шаг 3. Столбчатая по регионам (отсортированная)

```python
by_region = df.groupby("регион")["сумма"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(by_region.index, by_region.values)
plt.title("Продажи по регионам")
plt.xlabel("Регион")
plt.ylabel("Сумма, руб.")
plt.xticks(rotation=30)
plt.show()
```

## Шаг 4. Интерактивный график в Plotly

```python
import plotly.express as px

fig = px.line(monthly.reset_index(), x="месяц", y="сумма", title="Продажи по месяцам (интерактив)")
fig.show()
```
