## Шаг 1. Открываем PDF

```python
import fitz

doc = fitz.open("document.pdf")  # замените имя на своё
print("Страниц:", len(doc))
page = doc[0]
text = page.get_text()
print(text[:1500])
```

## Шаг 2. Извлекаем текст как Markdown (PyMuPDF4LLM)

```python
import pymupdf4llm

md = pymupdf4llm.to_markdown("document.pdf")
print(md[:2000])
```
