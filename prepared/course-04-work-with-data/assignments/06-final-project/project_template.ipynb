## 7. Экономика

Сколько занимает задача вручную, сколько — с агентом, стоит ли автоматизировать.

```python
hours_manual = 3      # часов в месяц вручную
hours_ai = 0.3        # часов в месяц с агентом
hour_cost = 500       # руб. за час
saved = (hours_manual - hours_ai) * hour_cost
print(f"Экономия в месяц: {saved:.0f} руб.")
print(f"Экономия за год: {saved * 12:.0f} руб.")
```
