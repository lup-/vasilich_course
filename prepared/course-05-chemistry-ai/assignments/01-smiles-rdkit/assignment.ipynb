## Шаг 1. Читаем SMILES

Запустите ячейку и посмотрите, что получается для разных строк.

```python
from rdkit import Chem

smiles_list = ["CCO", "CC=O", "CC(=O)O", "C1=CC=CC=C1"]
for s in smiles_list:
    m = Chem.MolFromSmiles(s)
    print(f"{s:15} -> {'OK' if m else 'ОШИБКА'}")
```

## Шаг 2. Канонический SMILES

RDKit умеет «приводить к каноническому виду» — одинаковые молекулы записываются одинаково.

```python
for s in smiles_list:
    m = Chem.MolFromSmiles(s)
    print(f"{s:15} -> {Chem.MolToSmiles(m)}")
```

## Шаг 3. Рисуем молекулы

Построим картинки. В Colab они появятся прямо в ноутбуке.

```python
from rdkit.Chem import Draw

mols = [Chem.MolFromSmiles(s) for s in smiles_list]
img = Draw.MolsToGridImage(mols, molsPerRow=2, subImgSize=(250, 200), legends=smiles_list)
img
```

## Шаг 4. Имя соединения и формула

Для этанола (`CCO`) выведем IUPAC-имя и брутто-формулу.

```python
from rdkit.Chem import rdMolDescriptors

mol = Chem.MolFromSmiles("CCO")
print("IUPAC:", Chem.MolToIUPACName(mol))
print("Формула:", rdMolDescriptors.CalcMolFormula(mol))
```
