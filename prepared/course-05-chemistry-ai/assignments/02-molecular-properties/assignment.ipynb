# Задание 02. Свойства молекул: расчёт и сравнение

Считаем дескрипторы через RDKit для набора известных веществ и сравниваем.

## Шаг 1. Набор молекул

Вода, этанол, глюкоза, аспирин, бензол — известные всем вещества.

```python
from rdkit import Chem

names = ["вода", "этанол", "глюкоза", "аспирин", "бензол"]
smiles = ["O", "CCO", "C(C1C(C(C(C(O1)O)O)O)O)O", "CC(=O)OC1=CC=CC=C1C(=O)O", "c1ccccc1"]
mols = [Chem.MolFromSmiles(s) for s in smiles]
print("Прочитано молекул:", sum(1 for m in mols if m))
```

## Шаг 2. Таблица дескрипторов

Масса, logP, TPSA, доноры/акцепторы H-связей.

```python
import pandas as pd
from rdkit.Chem import Descriptors

rows = []
for name, smi, m in zip(names, smiles, mols):
    if m is None:
        continue
    rows.append({
        "вещество": name,
        "SMILES": smi,
        "масса": Descriptors.MolWt(m),
        "logP": Descriptors.MolLogP(m),
        "TPSA": Descriptors.TPSA(m),
        "доноры H": Descriptors.NumHDonors(m),
        "акцепторы H": Descriptors.NumHAcceptors(m),
    })
df = pd.DataFrame(rows)
df
```

## Шаг 3. Сравнение

Какое вещество самое тяжёлое? У какого logP положительный (липофильный), у какого отрицательный (гидрофильный)?

```python
print("Самое тяжёлое:", df.loc[df["масса"].idxmax(), "вещество"])
print("Самое лёгкое:", df.loc[df["масса"].idxmin(), "вещество"])
print("Гидрофильные (logP < 0):", list(df[df["logP"] < 0]["вещество"]))
print("Липофильные (logP > 0):", list(df[df["logP"] > 0]["вещество"]))
```

## Шаг 4. Проверка: масса аспирина

Масса аспирина в PubChem — 180.16 г/моль. Сравните с расчётом RDKit.

```python
aspirin = df[df["вещество"] == "аспирин"]["масса"].values[0]
print(f"RDKit: {aspirin:.2f}  PubChem: 180.16  Совпадает: {abs(aspirin - 180.16) < 0.1}")
```
