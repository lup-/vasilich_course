## Шаг 1. Агент «Бухгалтер»

Запустите код ниже. Потребуется API-ключ (OPENAI_API_KEY) — вставьте его в переменные окружения, НЕ в код.

```python
from agents import Agent, Runner
import os

assert os.environ.get("OPENAI_API_KEY"), "Задайте OPENAI_API_KEY в переменных окружения"

def sum_sales(receipts: list[float]) -> float:
    """Считает сумму списка чисел (продаж)."""
    return sum(receipts)

agent = Agent(
    name="Бухгалтер",
    instructions="Ты считаешь суммы. Всегда вызывай инструмент sum_sales.",
    tools=[sum_sales],
)
result = Runner.run_sync(agent, "Посчитай сумму: 100, 250, 300")
print(result.final_output)
```
