## Шаг 1. Код сервера

Создайте файл `server.py` с содержимым ниже (можно попросить ИИ написать такой же и сверить).

```python
server_code = '''
from mcp.server.fastmcp import FastMCP

server = FastMCP("my-tools")

@server.tool()
def get_weather(city: str) -> str:
    """Возвращает строку с погодой для города (демо)."""
    return f"Погода в {city}: +18°C, солнечно"

@server.tool()
def sum_list(numbers: list[float]) -> float:
    """Считает сумму чисел."""
    return sum(numbers)

if __name__ == "__main__":
    server.run()
'''
with open("server.py", "w", encoding="utf-8") as f:
    f.write(server_code)
print("server.py создан")
```

## Шаг 2. Запуск сервера

Запустите сервер в отдельном терминале и проверьте, что он стартует без ошибок.

```python
print("Запуск: python server.py (в отдельном терминале)")
print("Сервер стартовал без ошибок?", "...")
```

## Шаг 3. Подключение к агенту

Добавьте в конфиг MCP клиента свой сервер (команда python, аргумент путь к server.py). Впишите конфиг.

```python
config = '''
{
  "mcpServers": {
    "my-tools": {
      "command": "python",
      "args": ["/path/to/server.py"]
    }
  }
}
'''
print("Мой конфиг:", config)
```
