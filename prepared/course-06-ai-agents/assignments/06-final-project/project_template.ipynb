## 5. Экономика

Вручную vs с агентом: часы, частота, стоимость часа, год.

```python
hours_manual = 1       # часов на задачу вручную
hours_ai = 0.2         # часов с агентом
times_per_week = 2
hour_cost = 600
weeks_per_year = 52

manual_year = hours_manual * times_per_week * weeks_per_year
ai_year = hours_ai * times_per_week * weeks_per_year
saving = (manual_year - ai_year) * hour_cost
print(f"Часов вручную за год: {manual_year:.0f}")
print(f"Часов с агентом за год: {ai_year:.0f}")
print(f"Экономия в год: {saving:,.0f} руб.")
```
