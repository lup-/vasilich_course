# Задание 03. Локальная модель из Python

Вызовем локальную модель через Ollama API из ноутбука.

```python
import requests

url = "http://localhost:11434/api/generate"
data = {
    "model": "qwen2.5:3b",
    "prompt": "Объясни, что такое локальная ИИ-модель, в трёх предложениях.",
    "stream": False,
}
try:
    resp = requests.post(url, json=data, timeout=120)
    if resp.ok:
        print(resp.json()["response"])
    else:
        print("Ошибка сервера:", resp.status_code)
except Exception as e:
    print("Не удалось связаться с Ollama:", e)
```

## Шаг 1. Функция ask_local

Напишем функцию-обёртку для вопросов.

```python
def ask_local(prompt, model="qwen2.5:3b"):
    import requests
    r = requests.post(
        "http://localhost:11434/api/generate",
        json={"model": model, "prompt": prompt, "stream": False},
        timeout=120,
    )
    if r.ok:
        return r.json()["response"].strip()
    return f"Ошибка: {r.status_code}"

print(ask_local("Назови столицу Японии."))
```

## Шаг 2. Несколько вопросов

Зададим цикл вопросов и соберём ответы в таблицу.

```python
import pandas as pd

questions = [
    "Сколько будет 2+2?",
    "Переведи на английский: «доброе утро».",
    "Назови три планеты Солнечной системы.",
]
answers = [ask_local(q) for q in questions]
df = pd.DataFrame({"вопрос": questions, "ответ": answers})
df
```
