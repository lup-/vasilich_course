## Шаг 1. Настройка окружения

Укажите локальный сервер и модель (переменные окружения).

```python
import os
os.environ["OPENAI_BASE_URL"] = "http://localhost:11434/v1"
os.environ["OPENAI_API_KEY"] = "ollama"
os.environ["OPENAI_MODEL"] = "qwen2.5:3b"
print("Окружение настроено на локальную модель")
```

## Шаг 2. Агент на локальной модели

Создадим агента и зададим вопрос.

```python
from agents import Agent, Runner

agent = Agent(
    name="Локальный помощник",
    instructions="Ты — помощник. Отвечай кратко и по делу.",
)
try:
    result = Runner.run_sync(agent, "Расскажи, кто ты и на какой модели работаешь.")
    print(result.final_output)
except Exception as e:
    print("Ошибка:", e)
```

## Шаг 3. Агент с инструментом

Добавим инструмент и проверим, что он вызывается.

```python
def sum_sales(receipts: list[float]) -> float:
    """Считает сумму списка чисел (продаж)."""
    return sum(receipts)

agent2 = Agent(
    name="Локальный бухгалтер",
    instructions="Ты считаешь суммы. Всегда вызывай инструмент sum_sales.",
    tools=[sum_sales],
)
try:
    r2 = Runner.run_sync(agent2, "Посчитай сумму: 100, 250, 300")
    print(r2.final_output)
except Exception as e:
    print("Ошибка:", e)
```
