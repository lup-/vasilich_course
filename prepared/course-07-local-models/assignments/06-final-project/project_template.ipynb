## 6. Экономика

Облако vs локально: расходы за год.

```python
requests_per_day = 30
days_per_month = 22
months_per_year = 12
cloud_cost_per_request = 0.5
local_power_month = 200
hardware_once = 0

cloud_year = requests_per_day * days_per_month * months_per_year * cloud_cost_per_request
local_year = local_power_month * months_per_year + hardware_once
print(f"Облако за год: {cloud_year:,.0f} руб.")
print(f"Локально за год: {local_year:,.0f} руб.")
print(f"Экономия: {cloud_year - local_year:,.0f} руб.")
```
